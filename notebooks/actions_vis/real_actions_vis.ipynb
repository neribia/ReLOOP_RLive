# Remote Actions Visualization

```python
import cv2

from rlive_env import RemoteWorldEnv, ActionSpaceType
from rlive_common.core.hardware_config import WorldConfig, CameraType, CameraResolution
from rlive_common.utils import get_logger

from vis_config import plot_image_grid, save_images, get_action_list, run_action_list

logger = get_logger(__name__)
```

```python
MAX_STEPS_PER_EPISODE = 400
N_RANDOM = 496
```

```python
def create_env():
    # Configure world settings
    world_config = WorldConfig(
        camera_type=CameraType.WEBCAM,
        camera_id=2,
        camera_resolution=CameraResolution.RES_640x480,
        bolt_name="BP-D217",
        bolt_use_dummy=False
    )

    return RemoteWorldEnv(
        max_episode_steps=MAX_STEPS_PER_EPISODE,
        base_url="http://127.0.0.1:8000",
        action_space_type=ActionSpaceType.CARTESIAN,
        world_config=world_config,
        decay=0,
    )
```

```python
import numpy as np
FIXED_ACTIONS = [
    np.array([-0.4, -0.6], dtype=np.float32),
    np.array([-1.0, -0.5], dtype=np.float32),
    np.array([0.5, -0.8], dtype=np.float32),
    np.array([-0.5, -0.5], dtype=np.float32),
    ]
env = create_env()
try:
    observation_list, action_list, rewards_list = run_action_list(env, get_action_list(env.action_space, n_random=N_RANDOM))
finally:
    env.close()
    del env
```

```python
cols = rows = 4
plot_image_grid(
    image_list=observation_list,
    label_list=action_list,
    reward_list=rewards_list,
    rows=rows,
    cols=cols,
)
```
