# Simple Action Visualization

```python
import os
# os.environ["DEBUG"] = "True"

from rlive_common import ActionSpaceType
from rlive_sim import (
    SimulationConfig,
    PhysicsConfig,
    RenderConfig,
    PhysicsBackend,
    RenderBackend,
    SimulationEnv,
)
from rlive_common.utils import get_logger

from vis_config import plot_image_grid, save_images, get_action_list, run_action_list

logger = get_logger(__name__)
```

```python
MAX_STEPS_PER_EPISODE = 600
N_RANDOM = 496
```

```python
def create_env():
    # Configure physics backend
    physics_config = PhysicsConfig(
        backend=PhysicsBackend.SIMPLE,
        # extra={
        #     "box_width": 0.8,
        #     "box_height": 0.4,
        #     "ball_radius": 0.07,
        # }
    )

    render_config = RenderConfig(
        backend=RenderBackend.OPENCV,
        width=640,
        height=480,
    )

    sim_config = SimulationConfig(
        use_integrated=False,
        physics=physics_config,
        render=render_config,
    )

    return SimulationEnv(
        config=sim_config,
        max_episode_steps=MAX_STEPS_PER_EPISODE,
        action_space_type=ActionSpaceType.CARTESIAN,
    )
```

```python
from rlive_sim import PhysicsState
import numpy as np
reset_options = {
    "initial_state": PhysicsState(
        position=[0.0, 0.0, 0.0],
        rotation=[0.0, 0.0, 90.0],
    )
}

env = create_env()
observation_list, action_list, rewards_list = run_action_list(
    env,
    get_action_list(env.action_space, n_random=N_RANDOM),
    reset_options=reset_options,
)
del env
```

```python
cols = rows = 4
plot_image_grid(
    image_list=observation_list,
    label_list=action_list,
    reward_list=rewards_list,
    rows=rows,
    cols=cols,
)
```

```python
save_images(observation_list, action_list, rewards_list, save_dir="image_folder/simple")
```
